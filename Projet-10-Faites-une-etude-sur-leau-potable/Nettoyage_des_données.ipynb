```python
#rassemblement des variables au même endroit dans une structure de dictionnaire Python (clé/valeur)
#permet d'automatiser des tâches sur l'ensemble des fichiers d'un seul coup avec un boucle
datasets = {
    "df_pop" : df_pop,
    "df_water"      : df_water,
    "df_mortality"  : df_mortality,
    "df_stability"  : df_stability,
    "df_geo"     : df_geo,
}
```

<div style="background-color: RGB(150, 20, 30);" >
<h2 style="margin: auto; padding: 20px; color:#fff; ">Etape 3 - Netoyage des données</h2>
</div>

<div style="border: 1px solid RGB(150, 20, 30);" >
<h3 style="margin: auto; padding: 20px; color: RGB(150, 20, 30); ">Nettoyage du fichier 'BasicAndSafelyManagedDrinkingWaterServices.csv'</h3>
</div>

```python
# --- Nettoyage ---
df_water_clean = df_water.copy()
print("📊 --- ÉTAT INITIAL ---")
inspector(df_water_clean)

#Renommage des colonnes
df_water_clean.columns = ['year', 'country', 'granularity', 'basic_pct', 'safely_managed_pct']
display(df_water_clean)

#On garde uniquement Total / Urban / Rural
df_water_clean = df_water_clean[df_water_clean['granularity'].isin(['Total', 'Urban', 'Rural'])]

#Enlève les éléments entre parenthèse dans la colonne 'country' pour que PowerBI reconnaisse les pays
df_water_clean['country'] = df_water_clean['country'].str.replace(r'\s*\(.*\)', '', regex=True)

df_water_clean['country'] = df_water_clean['country'].replace('Congo', 'Republic of Congo')

#Pivot : une ligne par pays/année, colonnes par granularité
df_water_pivot = df_water_clean.pivot_table(
    index=['country', 'year'],
    columns='granularity',
    values=['basic_pct', 'safely_managed_pct']
)
display(df_water_pivot)

#Aplatir les colonnes multi-niveaux
df_water_pivot.columns = [f"{val}_{gran.lower()}" for val, gran in df_water_pivot.columns]
df_water_pivot = df_water_pivot.reset_index()
display(df_water_pivot)

df_water_pivot['no_acces_pct']       = (100 -df_water_pivot['basic_pct_total']).round(2)
df_water_pivot['no_acces_rural_pct'] = (100 - df_water_pivot['basic_pct_rural']).round(2)
df_water_pivot['no_acces_urban_pct'] = (100 - df_water_pivot['basic_pct_urban']).round(2)

print("✅ Colonnes no_acces ajoutées")
print(f"   {df_water_pivot.shape[0]} lignes × {df_water_pivot.shape[1]} colonnes")
print(df_water_pivot[['country', 'year', 'no_acces_pct', 'no_acces_rural_pct', 'no_acces_urban_pct']].head(3))
display(df_water_pivot)

#On ajoute une colonne 'country_year' qui non permettra de faire les liaisons dans Power BI
df_water_pivot['country_year'] = df_water_pivot['country'] + '_' + df_water_pivot['year'].astype(str)

# --- Vérifier df ---
print("\n🎯 --- ÉTAT FINAL APRÈS NETTOYAGE & PIVOT ---")
inspector(df_water_pivot)

# --- Export fichier ---
df_water_pivot.to_csv("df_water_clean.csv", index=False, sep=';', decimal=',', encoding='utf-8-sig')

print("✅ Fichier exporté")
```

<div style="border: 1px solid RGB(150, 20, 30);" >
<h3 style="margin: auto; padding: 20px; color: RGB(150, 20, 30); ">Nettoyage du fichier 'MortalityRateAttributedToWater.csv'</h3>
</div>

```python
# --- Nettoyage ---
df_mortality_clean = df_mortality.copy()
print("📊 --- ÉTAT INITIAL ---")
inspector(df_mortality_clean)

df_mortality_clean = df_mortality_clean.rename(columns={
    'Year': 'year',
    'Country': 'country',
    'Granularity': 'granularity',
    'Mortality rate attributed to exposure to unsafe WASH services': 'wash_mortality_rate',
    'WASH deaths': 'wash_deaths'
})

#Enlève les éléments entre parenthèse dans la colonne 'country' pour que PowerBI reconnaisse les pays
df_mortality_clean['country'] = df_mortality_clean['country'].str.replace(r'\s*\(.*\)', '', regex=True)

df_mortality_clean['country'] = df_mortality_clean['country'].replace('Congo', 'Republic of Congo')

#Création d'une colonne par granularité
df_mortality_pivot = df_mortality_clean.pivot_table(
    index=['country', 'year'],
    columns='granularity',
    values=['wash_mortality_rate', 'wash_deaths'])

df_mortality_pivot.columns = [
    f"{col[0].lower()}_{col[1].lower()}" 
    for col in df_mortality_pivot.columns
]

df_mortality_pivot = df_mortality_pivot.reset_index()

#On ajoute une colonne 'country_year' qui non permettra de faire les liaisons dans Power BI
df_mortality_pivot['country_year'] = df_mortality_pivot['country'] + '_' + df_mortality_pivot['year'].astype(str)

# --- Vérifier df ---
print("\n🎯 --- ÉTAT FINAL APRÈS NETTOYAGE ---")
inspector(df_mortality_pivot)

# --- Export fichier ---
df_mortality_pivot.to_csv("df_mortality_clean.csv", index=False, sep=';', decimal=',', encoding='utf-8-sig')

print("✅ Fichier exporté")
```

<div style="border: 1px solid RGB(150, 20, 30);" >
<h3 style="margin: auto; padding: 20px; color: RGB(150, 20, 30); ">Nettoyage du fichier 'PoliticalStability.csv'</h3>
</div>

```python
# --- Nettoyage ---
df_stability_clean = df_stability.copy()
print("📊 --- ÉTAT INITIAL ---")
inspector(df_stability_clean)

#On garde uniquement 'Total'
df_stability_clean = df_stability_clean[df_stability_clean['Granularity'] == 'Total']

#On supprime la colonne 'Granularity' devenue inutile car il ne reste que 'Total'
df_stability_clean = df_stability_clean.drop(columns=['Granularity'])

#Renommage
df_stability_clean.columns = ['country', 'year', 'political_stability']


#Enlève les éléments entre parenthèse dans la colonne 'country' pour que PowerBI reconnaisse les pays
df_stability_clean['country'] = df_stability_clean['country'].str.replace(r'\s*\(.*\)', '', regex=True)
df_stability_clean['country'] = df_stability_clean['country'].replace('North Macedonia', 'Republic of North Macedonia')
df_stability_clean['country'] = df_stability_clean['country'].replace('Congo', 'Republic of Congo')

#On ajoute une colonne 'country_year' qui non permettra de faire les liaisons dans Power BI
df_stability_clean['country_year'] = df_stability_clean['country'] + '_' + df_stability_clean['year'].astype(str)


#Ajout de la colonne 'wash_mortality_rate_total'
df_stability_clean = pd.merge(
    df_stability_clean, df_mortality_pivot[['country_year', 'wash_mortality_rate_total']], on='country_year', how="left" ########### df_mortality_pivot
)
display(df_stability_clean)

# Ajout de la colonne 'basic_pct_total'
df_stability_clean = pd.merge(
    df_stability_clean, df_water_pivot[['country_year', 'basic_pct_total']], on='country_year', how="left"
)
display(df_stability_clean)

mask_2016 = df_stability_clean['year'] == 2016
val_min = df_stability_clean.loc[mask_2016, 'wash_mortality_rate_total'].min()
val_max = df_stability_clean.loc[mask_2016, 'wash_mortality_rate_total'].max()

mortality_norm = 1 - (df_stability_clean.loc[mask_2016, 'wash_mortality_rate_total'] - val_min) / (val_max - val_min)

df_stability_clean.loc[mask_2016, 'gov_score'] = (
    (mortality_norm + df_stability_clean.loc[mask_2016, 'basic_pct_total'] / 100) / 2
).round(4)

# --- Vérifier df ---
print("\n🎯 --- ÉTAT FINAL APRÈS NETTOYAGE ---")
inspector(df_stability_clean)

# --- Export fichier ---
df_stability_clean.to_csv("df_stability_clean.csv", index=False, sep=';', decimal=',', encoding='utf-8-sig')

print("✅ Fichier exporté")
```

```python
# --- Nettoyage ---
df_geo_clean = df_geo.copy()
print("📊 --- ÉTAT INITIAL ---")
inspector(df_geo_clean)

#Renommage
df_geo_clean.columns = ['region', 'country']

#Enlève les éléments entre parenthèse dans la colonne 'country' pour que PowerBI reconnaisse les pays
df_geo_clean['country'] = df_geo_clean['country'].str.replace(r'\s*\(.*\)', '', regex=True)

#Renomme 'Congo' pour qu'il soit identifier par PowerBI
df_geo_clean['country'] = df_geo_clean['country'].replace('Congo', 'Republic of Congo')

# --- Vérifier df ---
print("\n🎯 --- ÉTAT FINAL APRÈS NETTOYAGE ---")
inspector(df_geo_clean)

# --- Export fichier ---
df_geo_clean.to_csv("df_geo_clean.csv", index=False, encoding='utf-8-sig')

print("✅ Fichier exporté")
```

<h3 style="margin: auto; padding: 20px; color: RGB(150, 20, 30); ">Nettoyage du fichier 'Population.csv'</h3>
</div>

```python
# --- Nettoyage ---
df_pop_clean = df_pop.copy()
print("📊 --- ÉTAT INITIAL ---")
inspector(df_pop_clean)

#Renommage des colonnes
df_pop_clean.columns = ['country', 'granularity', 'year', 'population']

#Enlève les éléments entre parenthèse dans la colonne 'country' pour que PowerBI reconnaisse les pays
df_pop_clean['country'] = df_pop_clean['country'].str.replace(r'\s*\(.*\)', '', regex=True)

df_pop_clean['country'] = df_pop_clean['country'].replace('Congo', 'Republic of Congo')

#On multiplie la colone 'population' par 1000 pour avoir la notation française et ne pas avoir de point qui vont par la suite se transformer en virgule
df_pop_clean['population'] = df_pop_clean['population']*1000

#On garde uniquement Total / Urban / Rural / Male / Female
df_pop_clean = df_pop_clean[df_pop_clean['granularity'].isin(['Total', 'Urban', 'Rural', 'Male', 'Female'])]
display(df_pop_clean)

#Extraction des listes de pays uniques (en convertissant en "set" pour les opérations mathématiques)
countries_pop = set(df_pop_clean['country'].unique())
countries_water = set(df_water_pivot['country'].unique())

#Pays présents dans 'df_pop_clean' mais pas dans 'df_water_pivot'
only_in_pop = countries_pop - countries_water

#Pays présents dans 'df_water_pivot' mais pas dans 'df_pop_clean'
only_in_water = countries_water - countries_pop

# --- Affichage des résultats ---

print(f"Nombre de pays dans 'df_pop_clean' : {len(countries_pop)}")
print(f"Nombre de pays dans 'df_water_pivot' : {len(countries_water)}")
print("-" * 40)

print(f"\n🔴 Pays dans 'df_pop_clean' qui n'ont PAS de correspondance dans 'df_water_pivot' ({len(only_in_pop)}) :")
print(sorted(list(only_in_pop)))

print(f"\n🔵 Pays dans 'df_water_pivot' qui n'ont PAS de correspondance dans 'df_pop_clean' ({len(only_in_water)}) :")
print(sorted(list(only_in_water)))

df_pop_clean['country'] = df_pop_clean['country'].replace('North Macedonia', 'Republic of North Macedonia')


print("-----------------------------------------------------")
#Pour éviter que Power BI rate des jointures, on regarde si des pays de nos faits sont absents de la table df_geo_clean
all_countries_in_facts = set(
    pd.concat(
        [
            df_pop_clean["country"],
            df_water_clean["country"],
            df_mortality_clean["country"],
            df_stability_clean["country"],
        ]
    ).unique()
)
geo_countries = set(df_geo_clean["country"].unique())
missing_geo = all_countries_in_facts - geo_countries

if missing_geo:
    print(f"⚠️ Attention : {len(missing_geo)} pays présents dans les données n'ont pas de correspondance géographique.\n")
    print("Voici la liste des pays concernés :")
    
#On trie la liste par ordre alphabétique pour que ce soit plus lisible
    for country in sorted(missing_geo):
        print(f" - {country}")
else:
    print("✅ Parfait ! Tous les pays ont une correspondance dans la table géographique.")
print("-----------------------------------------------------")

df_pop_clean = df_pop_clean[df_pop_clean['country'].isin(geo_countries)]
display(df_pop_clean)

#Pivot : une ligne par pays/année
df_pop_pivot = df_pop_clean.pivot_table(
    index=['country', 'year'],
    columns='granularity',
    values='population'
).reset_index()
display(df_pop_pivot)

df_pop_pivot.columns = [
    f"pop_{col.lower()}" if col not in ['country', 'year'] else col 
    for col in df_pop_pivot.columns
]
display(df_pop_pivot)

#Fusion avec la table région
df_pop_pivot = df_pop_pivot.merge(df_geo_clean, on='country', how='left') 

#Calcul des pourcentages avec .div() pour éviter les erreurs si pop_total est nul ou NaN
df_pop_pivot['urban_pct'] = (df_pop_pivot['pop_urban'].div(df_pop_pivot['pop_total']) * 100).round(2).clip(upper=100)
df_pop_pivot['rural_pct'] = (df_pop_pivot['pop_rural'].div(df_pop_pivot['pop_total']) * 100).round(2).clip(upper=100)

#On ajoute une colonne 'country_year' qui non permettra de faire les liaisons dans Power BI
df_pop_pivot['country_year'] = df_pop_pivot['country'] + '_' + df_pop_pivot['year'].astype(str)

# --- Vérifier df ---
print("\n🎯 --- ÉTAT FINAL APRÈS NETTOYAGE ---")
inspector(df_pop_pivot)

# --- Export fichier ---
df_pop_pivot.to_csv("df_pop_clean.csv", index=False, sep=';', decimal=',', encoding='utf-8-sig')

print("✅ Fichier exporté")
```
