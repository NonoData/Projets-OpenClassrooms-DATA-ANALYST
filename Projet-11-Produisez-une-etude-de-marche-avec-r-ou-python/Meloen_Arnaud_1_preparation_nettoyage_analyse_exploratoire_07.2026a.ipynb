```python
# Pivot - On bascule les lignes du format "Long" au format "Large"

df_dispoali_2017_clean_pivot = df_dispoali_2017_clean.pivot_table(    ### p=production, x=exportation, m=importation
    index='Zone',                    # Une seule ligne par pays (Zone)
    columns='Élément',               # Les 3 éléments deviennent les 3 nouvelles colonnes
    values='Valeur',                 # On remplit avec les chiffres de la colonne Valeur
    aggfunc='sum'                    # Sécurité en cas de doublons (fait la somme)
).reset_index()                      # Permet de transformer 'Zone' en vraie colonne exploitable

# Nettoyage final du nom des colonnes
df_dispoali_2017_clean_pivot.columns.name = None

display(df_dispoali_2017_clean_pivot.head())

# Compte le nombre total de lignes contenant une parenthèse ouvrante ou fermante
nb_lignes = (
    df_dispoali_2017_clean["Zone"]
    .str.contains(r"[\(\)]", regex=True, na=False)
    .sum()
)
print(f"Il y a {nb_lignes} lignes avec des parenthèses.")

# Filtre les lignes qui contiennent des parenthèses dans la colonne Zone
mask = df_dispoali_2017_clean["Zone"].str.contains(
    r"[\(\)]", regex=True, na=False
)

# Extrait les valeurs uniques de ces lignes et on les affiche sous forme de liste
valeurs_uniques_parentheses = (
    df_dispoali_2017_clean.loc[mask, "Zone"].unique().tolist()
)

print(valeurs_uniques_parentheses)

# Compte le nombre total de lignes contenant une parenthèse ouvrante ou fermante
nb_lignes = (
    df_dispoali_2017_clean["Zone"]
    .str.contains(r"[\(\)]", regex=True, na=False)
    .sum()
)
print(f"Il y a {nb_lignes} lignes avec des parenthèses.")
```

```python
#Renommage des colonnes
df_dispoali_2017_clean_pivot.columns = ['Zone',
                                        'Disponibilité alimentaire en quantité (kg/personne/an)',
                                        'Disponibilité intérieure (en milliers de tonnes)',
                                        'Exportations - Quantité (en milliers de tonnes)',
                                        'Importations - Quantité (en milliers de tonnes)',
                                        'Production (en milliers de tonnes)']
display(df_dispoali_2017_clean_pivot)
```
